```python
import requests

# Links de teste: pedem só 5 linhas de cada API
urls = {
    "Pix por município": "https://olinda.bcb.gov.br/olinda/servico/Pix_DadosAbertos/versao/v1/odata/TransacoesPixPorMunicipio(DataBase=@DataBase)?@DataBase='202507'&$top=5&$format=json",
    "Fraudes Pix": "https://olinda.bcb.gov.br/olinda/servico/Pix_DadosAbertos/versao/v1/odata/EstatisticasFraudesPix(Database=@Database)?@Database='202507'&$top=5&$format=json",
    "IBGE população": "https://apisidra.ibge.gov.br/values/t/6579/n3/all/v/9324/p/last?formato=json",
}

for nome, url in urls.items():
    try:
        resposta = requests.get(url, timeout=20)
        print(f"{nome}: FUNCIONOU (código {resposta.status_code})")
        print(resposta.text[:200])
    except Exception as erro:
        print(f"{nome}: BLOQUEADO ({type(erro).__name__})")
    print("-" * 60)
```
